```python
#import pandas lib and read the csv file
import pandas as pd
df=pd.read_csv("windows_log.csv",index_col=False)
```

```python
#Displaying the first five rows of the csv file
df.head()
```

```python
#give the info of the dataframe
df.info()
```

```python
#gives the data regarding null values
df.isnull().sum()
```

```python
#removing the null data
df=df.dropna(subset='Task Category')
```

```python
df.columns
```

```python
# Checking the levels present
df['Level'].value_counts()
```

```python
#Extracting only the error and warning samples
df_extract=df[df['Level']!='Critical']
df_extract
```

```python
#Group by Task category as we have to find the the module and frequency
df_module=df_extract.groupby('Task Category').agg(
    {'Level':lambda x:' | '.join(x.unique()),
     'Date and Time':lambda a:' | '.join(a.astype(str).unique()),
     'Source':lambda y:' | '.join(y.unique()),
     'Event ID':lambda b:' | '.join(b.astype(str).unique())
    }
)
# Aggregate function is use to format the other columns as the result of the groupby give the repetiton in other columns.
```

```python
df_module
```

```python
#Getting the frequency or occurence and storing it as a new column
df['Task Category'].value_counts()
df_module['Frequency']=df['Task Category'].value_counts()
```

```python
df_module
```

```python
#Rearranging the frequency column
f_col=['Frequency']
other_col=[col for col in df_module.columns if col!='Frequency']
df_module=df_module[f_col+other_col]
df_module
```

```python
#Writing the dataframe to a csv file 
df_module.to_csv("Module_freq_processed.csv")
```
