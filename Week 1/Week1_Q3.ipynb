Importing the required libraries

```python
import pandas as pd
from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,ConfusionMatrixDisplay,classification_report
```

```python
#load the data 
iris=load_iris()
X=iris.data
y=iris.target
```

```python
#Split the dataset into training dataset and test dataset
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=10)
```

80% of the dataset is used for training and 20% for testing.
The random_state is the seed value. It is set so that every time the same random sequence is generated, without that each time the model is trained on different set of data.

```python
#Fitting the model
model=DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
#Prediction of the algorithm on the test dataset
y_pred=model.predict(X_test)
```

```python
#Evaluating the accuracy of the model
acc=accuracy_score(y_test,y_pred)
print(f'Accuracy :{acc:.2f}%')
```

```python
#Confusion Matrix
con_mat=confusion_matrix(y_test,y_pred)
disp=ConfusionMatrixDisplay(con_mat)
print('Confusion Matrix as array')
print(con_mat)
disp.plot()
```

The diagonal of the confusion matrix give the True Positive(correct prediction). While the rest of the column states regarding the false prediction

```python
#Classification report
report=classification_report(y_test,y_pred,output_dict=True)
df_report=pd.DataFrame(report)
df_report
```
