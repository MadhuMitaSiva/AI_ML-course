```python
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
import os
import numpy as np
```

```python
#loading the model
model = load_model('Defect_detection.keras')
```

```python
#loading single image for deployment
img_path='Manufacture_dataset/casting_data/deploy_data/cast_1.jpeg'
img = image.load_img(
    img_path,
    target_size=(224,224)
)
```

```python
img_arr= image.img_to_array(img)
img_arr.shape
```

```python
#Expand the dimension
image_array = np.expand_dims(img_arr, axis=0)
image_array.shape
```

```python
#Preprocess the input
image_array=preprocess_input(image_array)
```

```python
#prediction
predict = model.predict(image_array)
print(predict[0][0])
```

```python
def DefectDetection(img_arr,file_name):
    #Expanding the dimension by adding the batch size at the first position
    img_arr = np.expand_dims(img_arr, axis=0)
    #preprocess the image based on the model
    img_arr = preprocess_input(img_arr)
    #predict the image
    prediction = model.predict(img_arr,verbose=0)
    if (prediction[0][0] > 0.5):
        predicted_class = 'Defect'
    else:
        predicted_class = 'No Defect'
    print(f'File name: {file_name}')
    print(f'Predicted Class: {predicted_class}')
    print(f'Predicted Score: {prediction[0][0]}')
    print('-----'*10)

```

```python
# Prediction from the deploy_data image directory
main_fold = 'Manufacture_dataset'
sub_fold1 = 'casting_data'
sub_fold2 = 'deploy_data'
folder_path=os.path.join(main_fold,sub_fold1,sub_fold2)
for img_file in os.listdir(folder_path):
    img_path = os.path.join(folder_path,img_file)
    #loading the image
    img = image.load_img(img_path, target_size=(224,224))
    #Converting it to an array
    image_array = image.img_to_array(img)
    DefectDetection(image_array,img_file)
```
