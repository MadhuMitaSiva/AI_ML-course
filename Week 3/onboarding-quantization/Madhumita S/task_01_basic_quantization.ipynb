```python
import numpy as np
# Input tensor
tensor_val = np.array([1.0, -0.5, 3.2, -2.8, 0.0],dtype = np.float32)
# Scale and Zero point
scale = 0.025
zero_point = 0
```

1. q = round(x/scale)+zp
2. x = (q-zp)* scale
3. error = absolute difference

```python
# function to calculate quantized values
def quantized_tensor(tensor, scale, zero_point):
    scaled_tensor = tensor / scale
    rounded_tensor = np.round(scaled_tensor)
    shifted_tensor = rounded_tensor + zero_point
    clipped_tensor = np.clip(shifted_tensor, -128, 127)
    return clipped_tensor.astype(np.int8)
```

```python
# function to calculate dequantized vector
def dequantized_tensor(quantized_tensor, scale, zero_point):
    reshifted_tensor = quantized_tensor - zero_point
    rescaled_tensor = reshifted_tensor * scale
    return rescaled_tensor.astype(np.float32)
```

```python
#Quantization
quantize_tensor = quantized_tensor(tensor_val, scale, zero_point)
print("Original Tensor: ",tensor_val)
print("Quantized tensor: ",quantize_tensor)
```

```python
#Dequantized
dequantize_tensor = dequantized_tensor(quantize_tensor, scale, zero_point)
print("Original Tensor: ",tensor_val)
print("Dequantized tensor: ",dequantize_tensor)
```

```python
# Absolute error
abs_error = np.abs(tensor_val - dequantize_tensor)
print("Absolute error per element: ", abs_error)
```

```python
# Mean Absolute
mae = np.mean(abs_error)
print("Mean Absolute Error: ",mae)
```

```python
#Max Absolute error
max_abs_error = np.max(abs_error)
print("Maximum Absolute Error: ",max_abs_error)
```
